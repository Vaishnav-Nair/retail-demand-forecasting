```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train_df = pd.read_csv('../data/raw/train.csv', parse_dates=['Date'])
store_df = pd.read_csv('../data/raw/store.csv')

print(train_df.head(5))
print(store_df.head(5))
```

```python
print(train_df.shape)

clean_df = train_df.query("Open == 1 and Sales > 0")

print(clean_df.shape)
```

```python
df = pd.merge(clean_df, store_df, on = 'Store', how = 'left')

print(df.columns)
```

```python
df.isnull().sum()
```

```python
df['CompetitionDistance'] = df['CompetitionDistance'].fillna(999999)

df['CompetitionOpenSinceMonth'] = df['CompetitionOpenSinceMonth'].fillna(0)

df['CompetitionOpenSinceYear'] = df['CompetitionOpenSinceYear'].fillna(0)

df['Promo2SinceWeek'] = df['Promo2SinceWeek'].fillna(0)

df['Promo2SinceYear'] = df['Promo2SinceYear'].fillna(0)

df['PromoInterval'] = df['PromoInterval'].fillna('None')

df.isnull().sum()
```
